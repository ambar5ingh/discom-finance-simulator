```python
def check_baseline(df):
    sources = set(read_csv('sources.csv')['source_id'])
    missing = df[df['value'].isna()]['key'].tolist()
    bad_src = df[~df['source_id'].isin(sources)]['key'].tolist()
    return {'missing values': missing, 'rows without a valid source_id': bad_src}

check_baseline(read_csv('baseline_all_india_fy2021.csv'))
```
